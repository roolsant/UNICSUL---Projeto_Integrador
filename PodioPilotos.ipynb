##### importando as bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt
```

##### definindo o caminho do arquivo CSV

```python
pilotos_podio_csv = 'C:/Users/rodri/OneDrive - 5pwtcn/Cruzeiro_do_Sul_Virtual/Projeto_Integrador/Atividade_4/Pilotos_Podio.csv'
```

##### usando um bloco try-except para carregar o arquivo CSV

```python
try:
    df = pd.read_csv(pilotos_podio_csv, sep=';')
    print("Arquivo carregado com sucesso!")
    
except FileNotFoundError:
    print("Arquivo não encontrado. Verifique o caminho e tente novamente.") 
```

##### mostrando informações do DataFrame

```python
df.info()
```

##### criando um df com os pilotos que mais subiram ao pódio

```python
df_com_podios = df[df['Pódios'] > 0]
print("df_com_podios criado com sucesso!")
```

##### mostrando as primeiras linhas do DataFrame

```python
df_com_podios.head()
```

##### ordenando o df para uma melhor exibição no gráfico

```python
df_com_podios = df_com_podios.sort_values(by='Pódios', ascending=False)
print("df_com_podios ordenado com sucesso!")
```

##### visualizando o DataFrame ordenado

```python
display(df_com_podios)
```

#### criando grafico

```python
# definindo o tamanho da figura
plt.figure(figsize=(14, 8))

# criando o gráfico de barras e armazenando a referência das barras
bars = plt.bar(df_com_podios['Piloto'], df_com_podios['Pódios'], color='teal')

# adicionando os valores no topo de cada barra
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval, int(yval), va='bottom', ha='center', fontsize=10, fontweight='bold')

# adicionando título e rótulos
plt.title('Número de Pódios por Piloto na Temporada', fontsize=16)
plt.xlabel('Piloto', fontsize=12)
plt.ylabel('Quantidade de Pódios', fontsize=12)

# otimizando a visualização
plt.xticks(rotation=45, ha='right') # Gira os nomes dos pilotos para não sobrepor
plt.tight_layout() # Ajusta o layout para evitar cortes

# salvando o gráfico e exibe na tela
plt.savefig('grafico_podio_pilotos.png')
print("Gráfico salvo como 'grafico_podios_pilotos.png'")
plt.show()

```
